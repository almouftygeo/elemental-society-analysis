```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# Load your data
df = pd.read_excel('Major elements all.xlsx')

# Find actual column names
sio2_col = next((c for c in df.columns if 'sio2' in c.lower()), 'SiO2')
na2o_col = next((c for c in df.columns if 'na2o' in c.lower()), 'Na2O')
k2o_col = next((c for c in df.columns if 'k2o' in c.lower()), 'K2O')

print(f"Using columns: SiO2='{sio2_col}', Na2O='{na2o_col}', K2O='{k2o_col}'")

# ============================================================
# COMPLETE TAS DIAGRAM WITH CLASSIFICATION FIELDS
# ============================================================

fig, ax = plt.subplots(1, 1, figsize=(14, 10))

# Define TAS classification fields with colors and labels
# Format: (SiO2_range, Na2O+K2O range, color, label)
tas_fields = [
    # Basic/Ultrabasic rocks
    ((41, 45), (0, 3), '#d4e6f1', 'Picrobasalt'),
    ((41, 45), (3, 5), '#d5e8d4', 'Foidite'),
    ((41, 45), (5, 10), '#ffe6cc', 'Tephrite/\nBasanite'),
    ((41, 45), (10, 15), '#f8cecc', 'Phonolitic\nFoidite'),
    
    # Basalt region
    ((45, 52), (0, 5), '#aed6f1', 'Basalt'),
    ((45, 52), (5, 7), '#a9dfbf', 'Trachy-\nbasalt'),
    
    # Basaltic Andesite
    ((52, 57), (0, 5), '#85c1e9', 'Basaltic\nAndesite'),
    ((52, 57), (5, 7), '#7dcea0', 'Basaltic\nTrachy-andesite'),
    
    # Andesite
    ((57, 63), (0, 5), '#5dade2', 'Andesite'),
    ((57, 63), (5, 7), '#52be80', 'Trachy-\nandesite'),
    
    # Dacite
    ((63, 69), (0, 5), '#2e86c1', 'Dacite'),
    ((63, 69), (5, 7), '#28b463', 'Trachyte/\nTrachydacite'),
    
    # Rhyolite
    ((69, 77), (0, 8), '#1a5276', 'Rhyolite'),
    ((69, 77), (8, 15), '#d4ac0d', 'Alkali\nRhyolite'),
]

# Add colored polygons for each field
for (x_range, y_range, color, label) in tas_fields:
    x_min, x_max = x_range
    y_min, y_max = y_range
    
    # Create polygon vertices
    vertices = [
        (x_min, y_min),
        (x_max, y_min),
        (x_max, y_max),
        (x_min, y_max)
    ]
    
    polygon = Polygon(vertices, closed=True, 
                     facecolor=color, edgecolor='black', 
                     alpha=0.4, linewidth=1.5)
    ax.add_patch(polygon)
    
    # Add label at center of field
    x_center = (x_min + x_max) / 2
    y_center = (y_min + y_max) / 2
    ax.text(x_center, y_center, label, 
            ha='center', va='center', fontsize=8, fontweight='bold',
            color='black')

# Plot your data points
total_alkali = df[na2o_col] + df[k2o_col]
ax.scatter(df[sio2_col], total_alkali, 
           c='red', edgecolors='white', s=120, zorder=10, 
           label='Tokaj Samples', linewidth=2)

# Set labels and title
ax.set_xlabel('SiO₂ (wt%)', fontsize=14, fontweight='bold')
ax.set_ylabel('Na₂O + K₂O (wt%)', fontsize=14, fontweight='bold')
ax.set_title('TAS Classification – Tokaj Mountains (Total Alkali vs Silica)', 
             fontsize=16, fontweight='bold')

# Set axis limits
ax.set_xlim(40, 80)
ax.set_ylim(0, 15)

# Add grid
ax.grid(True, alpha=0.3, zorder=0)

# Add legend
ax.legend(loc='upper left', fontsize=12)

# Add annotation for classification zones
ax.text(42, 14, 'Ultrabasic\n& Basic', fontsize=10, style='italic', color='gray')
ax.text(52, 14, 'Intermediate', fontsize=10, style='italic', color='gray')
ax.text(65, 14, 'Acidic\n(Silicic)', fontsize=10, style='italic', color='gray')

plt.tight_layout()
plt.show()

# ============================================================
# CLASSIFY YOUR SAMPLES
# ============================================================
def classify_tas(sio2, total_alkali):
    """Classify rock type using simplified TAS boundaries"""
    if sio2 < 45:
        if total_alkali < 3: return 'Picrobasalt'
        elif total_alkali < 5: return 'Foidite'
        elif total_alkali < 10: return 'Tephrite/Basanite'
        else: return 'Phonolitic Foidite'
    elif sio2 < 52:
        if total_alkali < 5: return 'Basalt'
        else: return 'Trachybasalt'
    elif sio2 < 57:
        if total_alkali < 5: return 'Basaltic Andesite'
        else: return 'Basaltic Trachy-andesite'
    elif sio2 < 63:
        if total_alkali < 5: return 'Andesite'
        else: return 'Trachy-andesite'
    elif sio2 < 69:
        if total_alkali < 5: return 'Dacite'
        else: return 'Trachyte/Trachydacite'
    else:
        if total_alkali < 8: return 'Rhyolite'
        else: return 'Alkali Rhyolite'

# Apply classification
df['Rock_Type'] = [classify_tas(s, ta) for s, ta in zip(df[sio2_col], total_alkali)]

print("\n" + "=" * 60)
print("📊 TAS CLASSIFICATION RESULTS")
print("=" * 60)
print(f"\n{'Sample':<20} {'SiO2':>8} {'Na2O+K2O':>10} {'Rock Type':<25}")
print("-" * 65)
for idx, row in df.iterrows():
    sample_id = row.get('ID', row.get('Sample', idx))
    print(f"{str(sample_id):<20} {row[sio2_col]:>8.1f} {total_alkali[idx]:>10.1f} {row['Rock_Type']:<25}")

print("\n" + "=" * 60)
print("ROCK TYPE SUMMARY")
print("=" * 60)
rock_counts = df['Rock_Type'].value_counts()
for rock_type, count in rock_counts.items():
    print(f"  {rock_type:<25} : {count} samples")
```
