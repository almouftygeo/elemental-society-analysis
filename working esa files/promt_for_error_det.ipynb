```python
# After reading any file, make column names unique:
def make_columns_unique(df):
    cols = []
    seen = {}
    for c in df.columns:
        c_str = str(c).strip()
        if c_str in seen:
            seen[c_str] += 1
            cols.append(f"{c_str}_{seen[c_str]}")
        else:
            seen[c_str] = 0
            cols.append(c_str)
    df.columns = cols
    return df
```
