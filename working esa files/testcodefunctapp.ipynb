```python
# Test the improvement
import pandas as pd
from io import StringIO

def detect_delimiter(filepath):
    """NEW: Auto-detect file delimiter"""
    with open(filepath, 'r', encoding='latin1') as f:
        first_line = f.readline()
    
    # Count potential delimiters
    semicolons = first_line.count(';')
    commas = first_line.count(',')
    tabs = first_line.count('\t')
    
    if semicolons > commas and semicolons > tabs:
        return ';'
    elif commas > tabs:
        return ','
    else:
        return '\t'

# Test it
delimiter = detect_delimiter('all_new_data(all_new_data).csv')
print(f"Detected delimiter: '{delimiter}'")
```

```python
import sys
sys.path.append('.')
from elemental_society_gui import EnhancedGeochemistryEngine

# Test alteration indices
geo_engine = EnhancedGeochemistryEngine()

# Calculate CIA (Chemical Index of Alteration)
indices = geo_engine.calculate_alteration_indices(df, ['CIA', 'CIW'])
print("Alteration Indices:")
for k, v in indices.items():
    print(f"  {k}: {v}")

# Calculate element ratios
ratios = geo_engine.calculate_element_ratios(df, df.columns.tolist())
print("\nElement Ratios:")
for k, v in ratios.items():
    print(f"  {k}: {v}")

# Calculate grade statistics for economic elements
grades = geo_engine.calculate_grade_statistics(df, ['Cu', 'Zn', 'Pb', 'Au', 'Ag'])
print("\nGrade Statistics:")
for element, stats in grades.items():
    print(f"  {element}: Mean={stats.get('Mean', 'N/A')}")
```

```python
#quick visualizarion 
import matplotlib.pyplot as plt

# REE Pattern for the first 5 samples
ree_elements = ['La_ppm', 'Ce_ppm', 'Pr_ppm', 'Nd_ppm', 'Sm_ppm', 
                'Eu_ppm', 'Gd_ppm', 'Tb_ppm', 'Dy_ppm', 'Ho_ppm', 
                'Er_ppm', 'Tm_ppm', 'Yb_ppm', 'Lu_ppm']

plt.figure(figsize=(10, 6))
for i in range(min(5, len(df))):
    values = df.iloc[i][ree_elements].values
    plt.plot(range(len(ree_elements)), values, 'o-', label=df.iloc[i]['ID'])

plt.xticks(range(len(ree_elements)), [e.replace('_ppm', '') for e in ree_elements], rotation=45)
plt.xlabel('REE Elements')
plt.ylabel('Concentration (ppm)')
plt.title('REE Patterns - First 5 Samples')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
